```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

```python
model_name = "Zyphra/Zamba2-2.7B-Instruct"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, device_map="cuda", torch_dtype=torch.bfloat16)
```

```python
# input_text = "What factors contributed to the fall of the Roman Empire?"
# input_text = "In one season a flower blooms three times. In one year, there is one blooming season. How many times do two flowers bloom in two years? Please include your logic."
input_text = "user\nIn one season a flower blooms three times. In one year, there is one blooming season. How many times do two flowers bloom in two years? Please include your logic.\nassistant:\n"
input_ids = tokenizer(input_text, return_tensors="pt").to("cuda")

input_ids
```

```python
outputs = model.generate(**input_ids, max_new_tokens=5, num_beams=1, do_sample=False)
```

```python
print(f"{tokenizer.decode(outputs[0][input_ids['input_ids'].shape[1]:])!r}")
```
